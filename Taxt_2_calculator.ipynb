```python
class Calculator:

    def add(self, a, b):
        return a + b

    def subtract(self, a, b):
        return a - b

    def multiply(self, a, b):
        return a * b

    def divideInt(self, a, b):
        return a // b

    def divideFloat(self, a, b):
        return a / b

    def modulus(self, a, b):
        return a % b


calc = Calculator()

print("add =", calc.add(30, 10))
print("subtract=", calc.subtract(20, 2))
print("multiply", calc.multiply(8, 7))
print("divideint", calc.divideInt(20, 5))
print("dividefloat", calc.divideFloat(20, 5))
print("mpdulus", calc.modulus(20, 5))
```
